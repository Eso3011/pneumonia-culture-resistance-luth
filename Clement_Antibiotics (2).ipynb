```python
import numpy as np

antibiotics = df.loc[:, 'CEFTRIAXONE':'AZTREONAM'].columns

for col in antibiotics:
    df[col] = df[col].astype(str).str.strip()
    df[col] = df[col].replace('nan', np.nan)
```

```python
df['CIPROFLOXACIN'].unique()
```

```python
summary_list = []

for ab in antibiotics:
    counts = df[ab].value_counts()
    summary_list.append({
        'Antibiotic': ab,
        'Resistant': counts.get('R', 0),
        'Susceptible': counts.get('S', 0),
        'Intermediate': counts.get('I', 0)
    })

resistance_summary = pd.DataFrame(summary_list)
resistance_summary
```

```python
resistance_summary['Total'] = resistance_summary[['Resistant','Susceptible','Intermediate']].sum(axis=1)
resistance_summary['Resistant_%'] = (resistance_summary['Resistant']/resistance_summary['Total']*100).round(1)

resistance_summary
```

```python
def age_group(age):
    if age < 18:
        return "<18 years"
    elif age <= 40:
        return "18-40 years"
    elif age <= 60:
        return "41-60 years"
    else:
        return ">60 years"

df['Age_Group'] = df['Age_Numeric'].apply(age_group)
```

```python
from statsmodels.stats.proportion import proportion_confint
```

```python
ci_list = []

n_total = len(df)

for org, count in organism_freq.items():
    lower, upper = proportion_confint(count, n_total, method='wilson')
    ci_list.append({
        'Organism': org,
        'Count': count,
        'Percent': (count/n_total)*100,
        'CI_lower_%': lower*100,
        'CI_upper_%': upper*100
    })

prevalence_ci = pd.DataFrame(ci_list).round(2)
prevalence_ci
```
